```python
import pandas as pd

import matplotlib.pyplot as plt
```

```python
print("NETFLIX PROJECT")
df=pd.read_csv("netflix_data.csv")# import file
#clean data
print(df.dtypes) # type of data str/int64
print(df.columns) # all the column's
print(df.isnull().sum())
print(df.info()) # inforamtion about data 
```

```python
df=df.dropna(subset=['show_id', 'type', 'title', 'director', 'cast', 'country', 'date_added',
       'release_year', 'rating', 'duration', 'listed_in', 'description'])
```

```python
#type index and type.value's
type_count=df['type'].value_counts()
plt.Figure(figsize=(6,4))
plt.bar(type_count.index,type_count.values,color=['skyblue','orange'],label="TV show vs Movie's")
plt.xlabel("Type")# two type of data in column Type -> (TV show)&(Movie)
plt.ylabel("count")# Count the all data
plt.title("TV-Show's and Movie's")
plt.legend()
plt.tight_layout()
plt.savefig("bar_Tv-show_movies.png")

plt.show()
```

```python
#pie chart on rating
rating_count=df["rating"].value_counts()
plt.Figure(figsize=(6,5))
plt.pie(rating_count,labels=rating_count.index,autopct='%1.1f%%')
plt.title("Rating Count")
plt.tight_layout()
plt.savefig("pie_rating_count.png")
plt.show()
```

```python
#hist chart movie duration in minuts
movie_df=df[df['type']=='Movie'].copy()
movie_df["duration_int"] = movie_df['duration'].str.replace('min','').astype(int)
plt.Figure(figsize=(8,6))
plt.hist(movie_df['duration_int'],bins=30,color="purple",edgecolor='black')
plt.savefig("histchart.png")
plt.xlabel("duration minuts")
plt.ylabel("number of movies")
plt.title("duration's of movies")
plt.tight_layout()
plt.show()
```

```python
release_count = df["release_year"].value_counts().sort_index()

plt.figure(figsize=(10, 6))
plt.scatter(release_count.index, release_count.values, color="red", label="release year & shows")
plt.title("Release Year vs Number of Shows")
plt.xlabel("Release Year")
plt.ylabel("Number of Shows")
plt.legend()
plt.tight_layout()
plt.savefig("release_year.png")
plt.show()

```

```python
country_count=df["country"].value_counts().head(10)
plt.figure(figsize=(8,6))
plt.barh(country_count.index, country_count.values ,color="blue",label="most movies produces",edgecolor="black")
plt.xlabel("level of produce")
plt.ylabel("Country")
plt.title("which county make movie's in large amount")
plt.legend()
plt.tight_layout()
plt.savefig("barhchart.png")
plt.show()
```
